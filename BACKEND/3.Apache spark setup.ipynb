```python
from pyspark.sql import SparkSession
from pyspark.ml import Pipeline
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import RandomForestClassifier as SparkRF

# Initialize Spark session
spark = SparkSession.builder \
    .appName("MedicalDiagnosisEHR") \
    .config("spark.sql.adaptive.enabled", "true") \
    .getOrCreate()

# Real-time EHR processing pipeline
def process_ehr_data(ehr_stream):
    # Load EHR data into Spark DataFrame
    ehr_df = spark.read.format("json").load(ehr_stream)
    
    # Feature engineering for patient demographics and clinical history
    feature_cols = ['age', 'gender', 'blood_pressure', 'heart_rate', 'temperature']
    assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
    
    # Create ML pipeline for real-time validation
    rf_spark = SparkRF(featuresCol="features", labelCol="diagnosis")
    pipeline = Pipeline(stages=[assembler, rf_spark])
    
    return pipeline

# Pattern matching for ML prediction validation
def validate_ml_predictions(ml_predictions, ehr_patterns):
    # Cross-validate ML outputs with historical EHR patterns
    validation_scores = []
    
    for prediction in ml_predictions:
        similar_cases = ehr_patterns.filter(
            (ehr_patterns.symptoms == prediction['symptoms']) &
            (ehr_patterns.demographics == prediction['demographics'])
        ).count()
        
        confidence_boost = min(similar_cases * 0.1, 0.3)
        validation_scores.append(prediction['confidence'] + confidence_boost)
    
    return validation_scores
```
