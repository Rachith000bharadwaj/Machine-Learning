```python
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def comprehensive_model_validation(models, X_test, y_test):
    validation_results = {}
    
    for model_name, model in models.items():
        # Make predictions
        y_pred = model.predict(X_test)
        y_prob = model.predict_proba(X_test)
        
        # Calculate metrics
        accuracy = accuracy_score(y_test, y_pred)
        precision, recall, f1, support = precision_recall_fscore_support(y_test, y_pred, average='weighted')
        
        # Confidence analysis
        confidence_scores = np.max(y_prob, axis=1)
        avg_confidence = np.mean(confidence_scores)
        
        validation_results[model_name] = {
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'f1_score': f1,
            'avg_confidence': avg_confidence,
            'confusion_matrix': confusion_matrix(y_test, y_pred)
        }
    
    return validation_results

# Visualization for system performance monitoring
def create_performance_dashboard(validation_results):
    fig, axes = plt.subplots(2, 2, figsize=(15, 12))
    
    # Model comparison
    models = list(validation_results.keys())
    accuracies = [validation_results[model]['accuracy'] for model in models]
    
    axes[0,0].bar(models, accuracies)
    axes[0,0].set_title('Model Accuracy Comparison')
    axes[0,0].set_ylabel('Accuracy')
    
    # Confidence distribution
    confidences = [validation_results[model]['avg_confidence'] for model in models]
    axes[0,1].bar(models, confidences)
    axes[0,1].set_title('Average Confidence Scores')
    
    plt.tight_layout()
    return fig
```

```python
from datetime import datetime

# Incremental learning from user interactions
class IncrementalMedicalLearning:
    def __init__(self, base_model):
        self.model = base_model
        self.interaction_buffer = []
        self.update_threshold = 100  # Update after 100 new interactions
    
    def log_user_interaction(self, symptoms, actual_diagnosis, confidence):
        interaction = {
            'symptoms': symptoms,
            'diagnosis': actual_diagnosis,
            'confidence': confidence,
            'timestamp': datetime.now()
        }
        self.interaction_buffer.append(interaction)
        
        # Trigger model update if threshold reached
        if len(self.interaction_buffer) >= self.update_threshold:
            self.update_model()
    
    def update_model(self):
        # Prepare training data from interactions
        X_new = [interaction['symptoms'] for interaction in self.interaction_buffer]
        y_new = [interaction['diagnosis'] for interaction in self.interaction_buffer]
        
        # Incremental training
        self.model.partial_fit(X_new, y_new)
        
        # Clear buffer
        self.interaction_buffer = []
        
        print("Model updated with new user interactions")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
import matplotlib.pyplot as plt
import networkx as nx
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report

# --- New Imports from your added code ---
import pickle
import json
import os
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from transformers import AutoTokenizer, AutoModelForCausalLM

# --- New Imports for PySpark ---
from pyspark.sql import SparkSession
from pyspark.ml import Pipeline
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import RandomForestClassifier as SparkRF

# --- New Imports for PySpark Streaming ---
from pyspark.streaming import StreamingContext
from pyspark.sql.functions import udf, col, lit
from pyspark.sql.types import StringType
import json

# --- New Imports for Vosk/SQLite/Kivy ---
try:
    from vosk import Model, KaldiRecognizer
    import pyaudio
    VOSK_AVAILABLE = True
except ImportError:
    print("⚠️ vosk or pyaudio not installed. Speech recognition functions will not work.")
    VOSK_AVAILABLE = False
    
import sqlite3
from datetime import datetime

try:
    import os
    os.environ['KIVY_NO_ARGS'] = '1' # Prevent Kivy from parsing script args
    from kivy.app import App
    from kivy.uix.boxlayout import BoxLayout
    from kivy.uix.button import Button
    from kivy.uix.label import Label
    KIVY_AVAILABLE = True
except ImportError:
    print("⚠️ kivy not installed. GUI functions will not work.")
    KIVY_AVAILABLE = False

# --- New Imports for TFLite/Encryption ---
# TensorFlow is already imported
try:
    from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
    from cryptography.hazmat.primitives import padding
    from cryptography.hazmat.backends import default_backend
    from cryptography.hazmat.primitives import serialization
    from cryptography.hazmat.primitives.kdf.pbkdf2 import PBKDF2HMAC
    from cryptography.hazmat.primitives import hashes
    from cryptography.hazmat.primitives import keywrap
    CRYPTOGRAPHY_AVAILABLE = True
except ImportError:
    print("⚠️ cryptography not installed. Encryption functions will not work.")
    CRYPTOGRAPHY_AVAILABLE = False

# --- New Imports for Advanced Validation ---
from sklearn.metrics import precision_recall_fscore_support, confusion_matrix
import seaborn as sns


# ============================================================================
# SECTION 1: ORIGINAL SCIKIT-LEARN ENSEMBLE PIPELINE FUNCTIONS
# ============================================================================

def preprocess_medical_data(df, target_column='Disease'):
    """
    Preprocesses the medical data DataFrame.
    
    Args:
        df (pd.DataFrame): The raw DataFrame.
        target_column (str): The name of the target/label column to exclude from scaling.

    Returns:
        tuple: (processed_DataFrame, label_encoders_dict)
    """
    print("\n--- Preprocessing Data ---")
    df_processed = df.copy()
    label_encoders = {}
    
    # Handle missing values (simple fillna for this example)
    for col in df_processed.columns:
        if df_processed[col].dtype == 'object':
            df_processed[col] = df_processed[col].fillna('Missing')
        else:
            df_processed[col] = df_processed[col].fillna(0)
    
    # Encode categorical variables
    print("Encoding categorical features...")
    for column in df_processed.select_dtypes(include=['object']).columns:
        le = LabelEncoder()
        df_processed[column] = le.fit_transform(df_processed[column])
        label_encoders[column] = le
        print(f"  - Encoded '{column}'")
    
    # Feature scaling (excluding the target column)
    print("Scaling numeric features...")
    numeric_cols = df_processed.select_dtypes(include=[np.number]).columns
    
    # --- FIX: Exclude the target column from scaling ---
    if target_column in numeric_cols:
        # Need to check if target_column was encoded, if so it's a number
        target_col_name = target_column
    elif target_column in label_encoders:
        # The target column was text and is now encoded
        target_col_name = target_column
    else:
        target_col_name = None

    if target_col_name and target_col_name in numeric_cols:
        print(f"  - Excluding '{target_col_name}' from scaling.")
        numeric_cols = numeric_cols.drop(target_col_name)
    # --- End of Fix ---

    if not numeric_cols.empty:
        scaler = StandardScaler()
        df_processed[numeric_cols] = scaler.fit_transform(df_processed[numeric_cols])
        print(f"  - Scaled {len(numeric_cols)} columns.")
    else:
        print("  - No numeric columns to scale (after excluding target).")
    
    return df_processed, label_encoders

def plot_disease_distribution(df):
    """
    Creates a bar chart of the disease distribution.
    """
    plt.figure(figsize=(10, 6))
    df['Disease'].value_counts().plot(kind='bar', color='skyblue')
    plt.title('Distribution of Diseases in Mock Data')
    plt.xlabel('Disease')
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

def create_symptom_disease_graph(df):
    """
    Creates and visualizes a network graph of diseases and symptoms.
    Assumes symptom columns are named 'Symptom1', 'Symptom2', etc.
    """
    G = nx.Graph()
    
    symptom_cols = [col for col in df.columns if col.startswith('Symptom')]
    
    for _, row in df.iterrows():
        disease = row['Disease']
        if not G.has_node(disease):
            G.add_node(disease, type='disease')
            
        for col in symptom_cols:
            symptom = row[col]
            if pd.notna(symptom) and symptom != 'nan_placeholder':
                if not G.has_node(symptom):
                    G.add_node(symptom, type='symptom')
                G.add_edge(disease, symptom)
                
    plt.figure(figsize=(12, 12))
    
    # Separate nodes by type
    disease_nodes = [n for n, d in G.nodes(data=True) if d['type'] == 'disease']
    symptom_nodes = [n for n, d in G.nodes(data=True) if d['type'] == 'symptom']

    # Use a spring layout
    pos = nx.spring_layout(G, k=0.3, iterations=50)
    
    # Draw nodes
    nx.draw_networkx_nodes(G, pos, nodelist=disease_nodes, node_color='red', node_size=2000, label='Diseases')
    nx.draw_networkx_nodes(G, pos, nodelist=symptom_nodes, node_color='lightblue', node_size=500, label='Symptoms')
    
    # Draw edges
    nx.draw_networkx_edges(G, pos, alpha=0.5)
    
    # Draw labels
    nx.draw_networkx_labels(G, pos, font_size=8)
    
    plt.title("Symptom-Disease Network Graph")
    plt.axis('off')
    plt.show()

def plot_model_accuracies(accuracies):
    """
    Creates a bar chart comparing model accuracies.
    
    Args:
        accuracies (dict): A dictionary of {'model_name': accuracy_score}
    """
    if not accuracies:
        print("No accuracies to plot.")
        return

    names = list(accuracies.keys())
    values = list(accuracies.values())
    
    plt.figure(figsize=(10, 6))
    bars = plt.bar(names, values, color=['#FF5733', '#33FF57', '#3357FF', '#FF33A1', '#33FFF6'])
    
    # Add accuracy labels on top of bars
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2.0, yval, f'{yval*100:.2f}%', va='bottom', ha='center')

    plt.title('Model Accuracy Comparison')
    plt.xlabel('Model')
    plt.ylabel('Accuracy')
    plt.ylim(0, 1.1) # Set y-limit to 1.1 to make space for labels
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# Model ensemble for diagnosis prediction
def build_diagnosis_ensemble():
    models = {
        'random_forest': RandomForestClassifier(n_estimators=100, random_state=42),
        'gradient_boost': GradientBoostingClassifier(n_estimators=100, random_state=42),
        'naive_bayes': GaussianNB(),
        'svm': SVC(probability=True, random_state=42),
        'neural_network': MLPClassifier(hidden_layer_sizes=(100, 50), max_iter=500, random_state=42) # Increased max_iter
    }
    return models

def train_and_evaluate_models(models, X_train, y_train, X_test, y_test):
    """
    Trains and evaluates a dictionary of models.
    """
    accuracies = {}
    print("\n--- Training and Evaluating Ensemble Models ---")
    if X_train.shape[0] == 0:
        print("Error: Cannot train models with 0 samples.")
        return {}
        
    for name, model in models.items():
        print(f"Training {name}...")
        try:
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            acc = accuracy_score(y_test, y_pred)
            print(f"  - Accuracy: {acc*100:.2f}%")
            accuracies[name] = acc
        except Exception as e:
            print(f"  - FAILED to train {name}. Error: {e}")
            accuracies[name] = 0.0
            
    print("-----------------------------------------------")
    return accuracies

# Confidence scoring system
def calculate_confidence_scores(models, X_test):
    predictions = {}
    confidences = {}
    
    if X_test.shape[0] == 0:
        print("Cannot calculate confidences with 0 test samples.")
        return predictions, confidences
        
    for name, model in models.items():
        if not hasattr(model, "predict_proba"):
            print(f"  - Skipping confidences for {name} (no 'predict_proba' method).")
            continue
            
        try:
            prob = model.predict_proba(X_test)
            predictions[name] = model.predict(X_test)
            confidences[name] = np.max(prob, axis=1)
        except Exception as e:
            print(f"  - FAILED to get confidences for {name}. Error: {e}")

    return predictions, confidences


# ============================================================================
# SECTION 2: BIOMEDICAL (TRANSFORMERS) MODEL FUNCTIONS
# (Defined but not used by default in the main scripts)
# ============================================================================

def load_biomedical_model():
    model_name = "microsoft/biogpt"
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(model_name)
    return tokenizer, model

def extract_symptom_features(symptoms_text, tokenizer, model):
    inputs = tokenizer(symptoms_text, return_tensors="pt", padding=True, truncation=True)
    outputs = model(**inputs)
    # Extract embeddings for symptom feature representation
    return outputs.last_hidden_state.mean(dim=1)

# ============================================================================
# SECTION 3: KERAS DEEP LEARNING PIPELINE FUNCTIONS
# ============================================================================

# --- NEW GRAPHING FUNCTION FOR KERAS HISTORY ---
def plot_training_history(history):
    """
    Plots the training and validation accuracy/loss from a Keras history object.
    
    Args:
        history: A Keras History object from model.fit().
    """
    print("\n--- Plotting Keras Model Training History ---")
    
    # Create a figure with two subplots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10), sharex=True)
    
    # Plot Accuracy
    if 'accuracy' in history.history and 'val_accuracy' in history.history:
        ax1.plot(history.history['accuracy'], label='Training Accuracy', color='blue', linestyle='-')
        ax1.plot(history.history['val_accuracy'], label='Validation Accuracy', color='orange', linestyle='--')
        ax1.set_ylabel('Accuracy')
        ax1.set_title('Model Accuracy')
        ax1.legend(loc='lower right')
        ax1.grid(True)
    else:
        ax1.text(0.5, 0.5, 'Accuracy data not found in history', ha='center', va='center')
    
    # Plot Loss
    if 'loss' in history.history and 'val_loss' in history.history:
        ax2.plot(history.history['loss'], label='Training Loss', color='blue', linestyle='-')
        ax2.plot(history.history['val_loss'], label='Validation Loss', color='orange', linestyle='--')
        ax2.set_ylabel('Loss')
        ax2.set_xlabel('Epoch')
        ax2.set_title('Model Loss')
        ax2.legend(loc='upper right')
        ax2.grid(True)
    else:
        ax2.text(0.5, 0.5, 'Loss data not found in history', ha='center', va='center')
    
    plt.tight_layout()
    plt.show()

# --- Wrapper Function for your Keras Training Script ---
def run_keras_training_pipeline():
    """
    Runs the complete Keras training pipeline.
    This function expects a 'DATASETS/Training.csv' file.
    """
    print("\n" + "#"*70)
    print("STARTING: Keras Deep Learning Pipeline")
    print("#"*70)

    # ============================================================================
    # STEP 1: Load and Prepare Your Datasets
    # ============================================================================
    
    # NOTE: This path must be correct for this script to run.
    # It looks for a 'DATASETS' folder in the same directory as the script.
    BASE_PATH = "." 
    dataset_path = os.path.join(BASE_PATH, "DATASETS", "Training.csv")

    try:
        df_main = pd.read_csv(dataset_path)
    except FileNotFoundError:
        print(f"\n--- ERROR ---")
        print(f"Could not find file: {dataset_path}")
        print("Please make sure 'Training.csv' is inside a 'DATASETS' folder.")
        print("Skipping Keras pipeline.")
        print("#"*70 + "\n")
        return
        
    df = df_main.copy()

    print(f"\n✓ Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")

    # ============================================================================
    # STEP 2: Identify Disease and Symptom Columns
    # ============================================================================
    DISEASE_COLUMN = 'prognosis'  # Adjust if different

    # Get all symptom columns
    symptom_columns = [col for col in df.columns if col != DISEASE_COLUMN]

    if not symptom_columns:
        print(f"--- ERROR ---")
        print(f"No symptom columns found. Make sure '{DISEASE_COLUMN}' is the correct disease column name.")
        print("Skipping Keras pipeline.")
        return

    print(f"\n✓ Disease column: {DISEASE_COLUMN}")
    print(f"✓ Number of symptoms: {len(symptom_columns)}")
    print(f"✓ Number of unique diseases (before filtering): {df[DISEASE_COLUMN].nunique()}")

    # ============================================================================
    # STEP 2.5: FILTER OUT DISEASES WITH ONLY 1 SAMPLE
    # ============================================================================
    print("\n" + "="*60)
    print("Filtering diseases with insufficient samples...")
    print("="*60)

    # Count samples per disease
    disease_counts = df[DISEASE_COLUMN].value_counts()
    
    # Find diseases with only 1 sample
    single_sample_diseases = disease_counts[disease_counts == 1].index.tolist()

    if single_sample_diseases:
        print(f"\n⚠️  Found {len(single_sample_diseases)} diseases with only 1 sample:")
        for disease in single_sample_diseases[:5]:  # Show first 5
            print(f"    - {disease}")
        if len(single_sample_diseases) > 5:
            print(f"    ... and {len(single_sample_diseases) - 5} more")
        
        # Remove these diseases
        df = df[~df[DISEASE_COLUMN].isin(single_sample_diseases)]
        print(f"\n✓ Removed {len(single_sample_diseases)} diseases with single samples")
        print(f"✓ Remaining diseases: {df[DISEASE_COLUMN].nunique()}")
        print(f"✓ Remaining samples: {len(df)}")
    else:
        print("\n✓ All diseases have at least 2 samples")

    # ============================================================================
    # STEP 3: Create Symptom Vocabulary
    # ============================================================================
    symptom_vocabulary = {symptom: idx for idx, symptom in enumerate(symptom_columns)}

    print(f"\n✓ Created symptom vocabulary with {len(symptom_vocabulary)} symptoms")

    os.makedirs('models', exist_ok=True)
    with open('models/symptom_vocabulary.json', 'w') as f:
        json.dump(symptom_vocabulary, f, indent=2)
    print("✓ Saved symptom_vocabulary.json")

    # ============================================================================
    # STEP 4: Prepare Features (X) and Labels (y)
    # ============================================================================
    X = df[symptom_columns].values
    y = df[DISEASE_COLUMN].values

    print(f"\n✓ Feature matrix shape: {X.shape}")
    print(f"✓ Labels shape: {y.shape}")

    # ============================================================================
    # STEP 5: Encode Disease Labels
    # ============================================================================
    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)

    print(f"\n✓ Encoded {len(label_encoder.classes_)} disease classes")

    with open('models/label_encoder.pkl', 'wb') as f:
        pickle.dump(label_encoder, f)
    print("✓ Saved label_encoder.pkl")

    # ============================================================================
    # STEP 6: Split Data into Training and Testing Sets (FIXED)
    # ============================================================================
    print("\n" + "="*60)
    print("Splitting data...")
    print("="*60)

    # Check if we can use stratify
    disease_counts_final = df[DISEASE_COLUMN].value_counts()
    min_samples = disease_counts_final.min()
    print(f"\nMinimum samples per disease: {min_samples}")

    if min_samples >= 2:
        # Safe to use stratify
        print("✓ Using stratified split (maintains disease distribution)")
        X_train, X_test, y_train, y_test = train_test_split(
            X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
        )
    else:
        # Don't use stratify
        print("⚠️  Using regular split (some diseases still have <2 samples)")
        X_train, X_test, y_train, y_test = train_test_split(
            X, y_encoded, test_size=0.2, random_state=42
        )

    print(f"\n✓ Training set: {X_train.shape[0]} samples")
    print(f"✓ Testing set: {X_test.shape[0]} samples")

    # ============================================================================
    # STEP 7: Build Neural Network Model
    # ============================================================================
    num_features = X_train.shape[1]
    num_classes = len(label_encoder.classes_)

    print(f"\n✓ Building model with {num_features} input features and {num_classes} output classes")

    model = Sequential([
        Dense(256, activation='relu', input_shape=(num_features,)),
        Dropout(0.3),
        Dense(128, activation='relu'),
        Dropout(0.3),
        Dense(64, activation='relu'),
        Dropout(0.2),
        Dense(num_classes, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    print("\n✓ Model architecture:")
    model.summary()

    # ============================================================================
    # STEP 8: Train the Model
    # ============================================================================
    print("\n" + "="*60)
    print("Starting Training...")
    print("="*60)

    history = model.fit(
        X_train, y_train,
        validation_data=(X_test, y_test),
        epochs=50,
        batch_size=32,
        verbose=1
    )
    
    # --- HERE IS THE NEW GRAPH CODE ---
    # Call the plotting function with the history object
    plot_training_history(history)
    
    # ============================================================================
    # STEP 9: Evaluate the Model
    # ============================================================================
    print("\n" + "="*60)
    print("Model Evaluation")
    print("="*60)

    train_loss, train_accuracy = model.evaluate(X_train, y_train, verbose=0)
    test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=0)

    print(f"\n✓ Training Accuracy: {train_accuracy*100:.2f}%")
    print(f"✓ Testing Accuracy: {test_accuracy*100:.2f}%")

    # ============================================================================
    # STEP 10: Save the Model
    # ============================================================================
    model.save('models/ml_model.h5')
    print("\n✓ Model saved as 'models/ml_model.h5'")
    
    # (Steps 11 and 12 from your script are omitted for brevity, but would be here)
    # ... (Sample predictions and saving summary) ...

    print("\n" + "="*60)
    print("Keras Training Complete!")
    print("="*60)
    print(f"\n✓ Keras Model can diagnose {len(label_encoder.classes_)} diseases")
    print(f"✓ Keras Training accuracy: {train_accuracy*100:.2f}%")
    print(f"✓ Keras Testing accuracy: {test_accuracy*100:.2f}%")
    print("#"*70 + "\n")


# ============================================================================
# SECTION 4: PYSPARK EHR PROCESSING FUNCTIONS
# ============================================================================

# Initialize Spark session
# Note: This will initialize when the script is imported.
# For a production app, you might want this inside a function.
try:
    spark = SparkSession.builder \
        .appName("MedicalDiagnosisEHR") \
        .config("spark.sql.adaptive.enabled", "true") \
        .getOrCreate()
    print("\n✓ SparkSession initialized.")
except Exception as e:
    print(f"\n⚠️ Could not initialize SparkSession. PySpark functions will not work.")
    print(f"   Error: {e}")
    spark = None

# Real-time EHR processing pipeline
def process_ehr_data(ehr_stream):
    if not spark:
        print("SparkSession not available. Cannot process EHR data.")
        return None
        
    # Load EHR data into Spark DataFrame
    ehr_df = spark.read.format("json").load(ehr_stream)
    
    # Feature engineering for patient demographics and clinical history
    feature_cols = ['age', 'gender', 'blood_pressure', 'heart_rate', 'temperature']
    assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
    
    # Create ML pipeline for real-time validation
    rf_spark = SparkRF(featuresCol="features", labelCol="diagnosis")
    pipeline = Pipeline(stages=[assembler, rf_spark])
    
    return pipeline

# Pattern matching for ML prediction validation
def validate_ml_predictions(ml_predictions, ehr_patterns):
    # Cross-validate ML outputs with historical EHR patterns
    validation_scores = []
    
    for prediction in ml_predictions:
        # This assumes ehr_patterns is a Spark DataFrame
        try:
            similar_cases = ehr_patterns.filter(
                (ehr_patterns.symptoms == prediction['symptoms']) &
                (ehr_patterns.demographics == prediction['demographics'])
            ).count()
            
            confidence_boost = min(similar_cases * 0.1, 0.3)
            validation_scores.append(prediction['confidence'] + confidence_boost)
        except Exception as e:
            print(f"Error during validation: {e}")
            validation_scores.append(prediction['confidence']) # Append original confidence
    
    return validation_scores

# ============================================================================
# SECTION 5: PYSPARK STREAMING DIAGNOSIS FUNCTIONS
# ============================================================================

# --- This is the logic ported from app.py's demo_diagnosis ---
def get_demo_diagnosis_logic(symptoms):
    """
    This function contains the rule-based logic from your app.py's
    demo_diagnosis function.
    """
    if not symptoms:
        return "No symptoms provided"
        
    symptoms_lower = symptoms.lower()
    
    if 'fever' in symptoms_lower and 'cough' in symptoms_lower:
        if 'chest pain' in symptoms_lower or 'difficulty breathing' in symptoms_lower:
            return 'Acute Bronchitis'
        elif 'body aches' in symptoms_lower or 'chills' in symptoms_lower:
            return 'Influenza'
        else:
            return 'Common Cold'
    elif 'headache' in symptoms_lower and ('fever' in symptoms_lower or 'nausea' in symptoms_lower):
        return 'Migraine'
    elif 'nausea' in symptoms_lower or 'vomiting' in symptoms_lower:
        if 'diarrhea' in symptoms_lower or 'abdominal pain' in symptoms_lower:
            return 'Gastroenteritis'
        else:
            return 'Indigestion'
    elif 'rash' in symptoms_lower or 'itching' in symptoms_lower:
        return 'Allergic Reaction'
    elif 'sore throat' in symptoms_lower and 'fever' in symptoms_lower:
        return 'Strep Throat'
    elif 'joint pain' in symptoms_lower and 'fatigue' in symptoms_lower:
        return 'Viral Arthritis'
    elif 'runny nose' in symptoms_lower and 'sneezing' in symptoms_lower:
        return 'Allergic Rhinitis'
    else:
        return 'Upper Respiratory Infection'
# --- End of ported logic ---

# 1. Register the UDF (User-Defined Function)
# This tells Spark how to use our Python function in a distributed way
if spark: # Only register if spark initialized successfully
    diagnosis_udf = udf(get_demo_diagnosis_logic, StringType())
else:
    diagnosis_udf = None

# 2. Define the missing function that was causing the error
def apply_diagnosis_models(patient_df):
    """
    Applies the diagnosis logic to the patient DataFrame using a UDF.
    Assumes the DataFrame 'patient_df' has a column named 'symptoms'.
    """
    if not diagnosis_udf:
        print("Error: diagnosis_udf not registered because Spark failed to initialize.")
        return patient_df.withColumn("prediction", lit("Error: Spark not available"))

    # Check if the required column 'symptoms' exists
    if "symptoms" not in patient_df.columns:
        print("Error: DataFrame does not have a 'symptoms' column.")
        # Return the DataFrame with an error message
        return patient_df.withColumn("prediction", lit("Error: Missing symptoms column"))

    # 3. Apply the UDF to create a new "prediction" column
    predictions_df = patient_df.withColumn(
        "prediction",
        diagnosis_udf(col("symptoms"))
    )
    return predictions_df

# Real-time streaming for continuous EHR monitoring
def setup_streaming_context():
    """
    Sets up the Spark Streaming context.
    Note: This function doesn't start the stream.
    """
    if not spark:
        print("SparkSession not available. Cannot set up streaming context.")
        return None
        
    # Use the existing SparkContext from the 'spark' session
    ssc = StreamingContext(spark.sparkContext, 10)  # 10-second batches
    
    # Connect to EHR data streams
    ehr_stream = ssc.socketTextStream("localhost", 9999)
    
    # Process incoming patient data in real-time
    def process_patient_data(rdd):
        if not rdd.isEmpty():
            # Ensure spark session is available
            global spark
            if not spark:
                print("SparkSession lost. Cannot process RDD.")
                return
                
            try:
                # Assuming the RDD contains JSON strings like '{"symptoms": "fever and cough"}'
                patient_df = spark.read.json(rdd)
                
                # Apply ML models and validation
                # This function is now defined and will work!
                predictions = apply_diagnosis_models(patient_df)
                
                # You should do something with the predictions here,
                # like save them to a database or another stream.
                # For this example, we'll just show them in the console.
                print("--- Batch Predictions ---")
                predictions.show()
                
                return predictions
            except Exception as e:
                print(f"Error processing RDD: {e}")
    
    ehr_stream.foreachRDD(process_patient_data)
    return ssc

# ============================================================================
# SECTION 6: NEW GRAPHING FUNCTION FOR STREAMING RESULTS
# ============================================================================

def plot_streaming_results(predictions_list):
    """
    Plots the count of diagnoses from a list of streaming predictions.
    
    Args:
        predictions_list (list): A list of prediction strings (e.g., ['Flu', 'Common Cold', 'Flu'])
    """
    if not predictions_list:
        print("No streaming predictions to plot.")
        return
        
    print("\n--- Plotting Simulated Streaming Results ---")
    
    # Convert the list of predictions to a Pandas Series to easily count values
    df = pd.Series(predictions_list, name="Diagnosis")
    
    plt.figure(figsize=(12, 7))
    df.value_counts().plot(kind='bar', color='lightgreen')
    plt.title('Distribution of Diagnoses from Stream (Sample)')
    plt.xlabel('Diagnosed Disease')
    plt.ylabel('Frequency (Count)')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# ============================================================================
# SECTION 7: VOSK, SQLITE, AND KIVY INTEGRATION
# ============================================================================

# --- Vosk Speech Recognition ---
def recognize_speech():
    if not VOSK_AVAILABLE:
        print("Vosk/PyAudio not found. Cannot recognize speech.")
        print("Simulating speech: 'fever and cough'")
        return "fever and cough"
        
    try:
        # Assumes "vosk-model-small-en-us-0.15" is in the same directory
        model = Model("vosk-model-small-en-us-0.15") 
    except Exception as e:
        print(f"Error loading Vosk model: {e}")
        print("Please download 'vosk-model-small-en-us-0.15' and unzip it.")
        print("Simulating speech: 'fever and cough'")
        return "fever and cough"

    recognizer = KaldiRecognizer(model, 16000)
    audio = pyaudio.PyAudio()
    stream = audio.open(format=pyaudio.paInt16, channels=1, rate=16000,
                        input=True, frames_per_buffer=8192)
    stream.start_stream()
    print("Speak...")
    
    text = ""
    while True:
        data = stream.read(4096)
        if recognizer.AcceptWaveform(data):
            text = json.loads(recognizer.Result())["text"]
            if text: # Break on first confirmed result
                print("You said:", text)
                break
    
    print("Speech recognition complete.")
    stream.stop_stream()
    stream.close()
    audio.terminate()
    return text

# --- SQLite Database Logging ---
try:
    conn = sqlite3.connect('medical_diagnosis.db')
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS symptoms (
            id INTEGER PRIMARY KEY,
            name TEXT,
            severity INTEGER)
    ''')
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS diagnoses (
            id INTEGER PRIMARY KEY,
            symptom_ids TEXT,
            predicted_disease TEXT,
            confidence REAL,
            timestamp LONG)
    ''')
    print("\n✓ SQLite database 'medical_diagnosis.db' initialized.")
except Exception as e:
    print(f"⚠️ Could not initialize SQLite database. Error: {e}")
    conn = None
    cursor = None

def log_diagnosis_interaction(symptoms, diagnosis, confidence):
    if not conn or not cursor:
        print("Database not available. Skipping log.")
        return
        
    sym_str = ",".join(symptoms)
    ts = int(datetime.now().timestamp())
    
    try:
        cursor.execute('''
            INSERT INTO diagnoses (symptom_ids, predicted_disease, confidence, timestamp)
            VALUES (?, ?, ?, ?)
        ''', (sym_str, diagnosis, confidence, ts))
        conn.commit()
        print(f"✓ Logged diagnosis: {diagnosis}")
    except Exception as e:
        print(f"Error logging to database: {e}")
        
# Example usage (and test):
print("Testing database log...")
log_diagnosis_interaction(["fever", "cough"], "Flu", 0.92)
if cursor:
    try:
        cursor.execute('SELECT * FROM diagnoses ORDER BY id DESC LIMIT 1')
        print(f"  - Last DB entry: {cursor.fetchone()}")
    except Exception as e:
        print(f"  - Could not fetch from DB: {e}")


# --- Kivy GUI Application ---
if KIVY_AVAILABLE:
    class MainWidget(BoxLayout):
        def __init__(self, **kwargs):
            super().__init__(**kwargs)
            self.orientation = 'vertical'
            self.label = Label(text="Medical Diagnosis App")
            self.add_widget(self.label)
            self.button = Button(text="Run Voice Diagnosis")
            self.button.bind(on_press=self.run_diagnosis)
            self.add_widget(self.button)


        def run_diagnosis(self, instance):
            # This would trigger vosk recognition & prediction
            # For a real app, you'd call recognize_speech() here,
            # then feed the text to a prediction model.
            self.label.text = "Listening... (Speak now)"
            
            # --- Simulation ---
            # In a real Kivy app, you'd run speech in a separate thread
            # to avoid freezing the GUI.
            # simulated_symptoms = recognize_speech() 
            simulated_symptoms = "fever and cough"
            simulated_diagnosis = "Flu"
            simulated_confidence = 0.92
            
            self.label.text = f"You said: {simulated_symptoms}\nDiagnosis: {simulated_diagnosis} (conf. {simulated_confidence:.2f})"
            
            # Log the diagnosis
            log_diagnosis_interaction(simulated_symptoms.split(), simulated_diagnosis, simulated_confidence)


    class MedicalDiagnosisApp(App):
        def build(self):
            return MainWidget()
else:
    print("Kivy classes not defined as Kivy is not available.")

# --- To run the Kivy app ---
# if __name__ == "__main__" and KIVY_AVAILABLE:
#     print("Starting Kivy GUI Application...")
#     MedicalDiagnosisApp().run()
#
# NOTE: Kivy app is not run by default in this script to allow
#       the other pipelines to execute in the console.

# ============================================================================
# SECTION 8: TFLITE CONVERSION AND ENCRYPTION FUNCTIONS
# ============================================================================

# --- TFLite Conversion ---
def convert_to_tflite(model_path, output_path):
    """
    Converts a saved Keras model to a TFLite model.
    """
    print(f"\n--- Converting model {model_path} to TFLite ---")
    try:
        # Load trained model
        model = tf.keras.models.load_model(model_path)
        
        # Convert to TensorFlow Lite
        converter = tf.lite.TFLiteConverter.from_keras_model(model)
        converter.optimizations = [tf.lite.Optimize.DEFAULT]
        converter.target_spec.supported_types = [tf.lite.constants.FLOAT16]
        
        tflite_model = converter.convert()
        
        # Save optimized model
        with open(output_path, 'wb') as f:
            f.write(tflite_model)
        
        print(f"✓ Successfully converted model to {output_path}")
        return output_path
    except Exception as e:
        print(f"⚠️ TFLite conversion FAILED. Error: {e}")
        return None

# Quantization for mobile optimization
# Note: 'representative_dataset_generator' is not defined in this script.
# This function is here as a template.
def quantize_model_for_mobile(model_path, representative_dataset_generator):
    """
    Quantizes a TFLite model for mobile optimization.
    Requires a 'representative_dataset_generator'.
    """
    print(f"\n--- Quantizing model {model_path} ---")
    try:
        converter = tf.lite.TFLiteConverter.from_saved_model(model_path)
        converter.optimizations = [tf.lite.Optimize.OPTIMIZE_FOR_SIZE]
        converter.representative_dataset = representative_dataset_generator
        converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
        
        quantized_model = converter.convert()
        print(f"✓ Model quantized successfully.")
        return quantized_model
    except Exception as e:
        print(f"⚠️ Model quantization FAILED. Error: {e}")
        return None

# --- Data Encryption (AES-GCM) ---
if CRYPTOGRAPHY_AVAILABLE:
    def generate_secret_key():
        # 256-bit AES key
        key = os.urandom(32)  # Returns bytes object, keep safe!
        return key

    def encrypt_data(key, plaintext):
        # AES-GCM parameters
        iv = os.urandom(12)  # GCM standard IV length is 12
        encryptor = Cipher(
            algorithms.AES(key),
            modes.GCM(iv),
            backend=default_backend()
        ).encryptor()
        # Encrypt
        ciphertext = encryptor.update(plaintext.encode()) + encryptor.finalize()
        tag = encryptor.tag
        return iv, ciphertext, tag

    def decrypt_data(key, iv, ciphertext, tag):
        decryptor = Cipher(
            algorithms.AES(key),
            modes.GCM(iv, tag),
            backend=default_backend()
        ).decryptor()
        # Decrypt
        try:
            plaintext = decryptor.update(ciphertext) + decryptor.finalize()
            return plaintext.decode()
        except Exception as e:
            print(f"Decryption FAILED: {e}")
            return None
else:
    print("Cryptography library not loaded. Encryption functions are disabled.")
    # Create dummy functions
    def generate_secret_key():
        print("Error: Cryptography not available.")
        return None
    def encrypt_data(key, plaintext):
        print("Error: Cryptography not available.")
        return None, None, None
    def decrypt_data(key, iv, ciphertext, tag):
        print("Error: Cryptography not available.")
        return None

# ============================================================================
# SECTION 9: ADVANCED VALIDATION & INCREMENTAL LEARNING (NEW)
# ============================================================================

def comprehensive_model_validation(models, X_test, y_test):
    """
    Runs a comprehensive validation on all models, including metrics
    and confidence scores.
    """
    validation_results = {}
    
    for model_name, model in models.items():
        print(f"  - Validating {model_name}...")
        # Make predictions
        y_pred = model.predict(X_test)
        
        # --- Robust predict_proba ---
        y_prob = None
        avg_confidence = 0.0
        confidence_scores = []
        if hasattr(model, "predict_proba"):
            try:
                y_prob = model.predict_proba(X_test)
                confidence_scores = np.max(y_prob, axis=1)
                avg_confidence = np.mean(confidence_scores)
            except Exception as e:
                print(f"    - Warning: Could not get predict_proba for {model_name}. Confidence will be 0. Error: {e}")
        else:
            print(f"    - Info: {model_name} has no predict_proba. Confidence will be 0.")
        # --- End of robust block ---
        
        # Calculate metrics
        accuracy = accuracy_score(y_test, y_pred)
        precision, recall, f1, support = precision_recall_fscore_support(y_test, y_pred, average='weighted', zero_division=0)
        
        validation_results[model_name] = {
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'f1_score': f1,
            'avg_confidence': avg_confidence,
            'confusion_matrix': confusion_matrix(y_test, y_pred)
        }
    
    return validation_results

def create_performance_dashboard(validation_results):
    """
    Creates a 2x2 dashboard of model performance metrics.
    """
    fig, axes = plt.subplots(2, 2, figsize=(15, 12))
    fig.suptitle('Model Performance Dashboard', fontsize=16)
    
    models = list(validation_results.keys())
    
    # 1. Accuracy
    accuracies = [validation_results[model]['accuracy'] for model in models]
    axes[0,0].bar(models, accuracies, color='blue')
    axes[0,0].set_title('Model Accuracy Comparison')
    axes[0,0].set_ylabel('Accuracy')
    axes[0,0].tick_params(axis='x', rotation=45)
    
    # 2. Average Confidence
    confidences = [validation_results[model]['avg_confidence'] for model in models]
    axes[0,1].bar(models, confidences, color='green')
    axes[0,1].set_title('Average Confidence Scores')
    axes[0,1].set_ylabel('Avg. Confidence')
    axes[0,1].tick_params(axis='x', rotation=45)

    # 3. F1-Score
    f1_scores = [validation_results[model]['f1_score'] for model in models]
    axes[1,0].bar(models, f1_scores, color='red')
    axes[1,0].set_title('Weighted F1-Score')
    axes[1,0].set_ylabel('F1-Score')
    axes[1,0].tick_params(axis='x', rotation=45)
    
    # 4. Recall
    recalls = [validation_results[model]['recall'] for model in models]
    axes[1,1].bar(models, recalls, color='purple')
    axes[1,1].set_title('Weighted Recall')
    axes[1,1].set_ylabel('Recall')
    axes[1,1].tick_params(axis='x', rotation=45)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust for suptitle
    return fig

def plot_confusion_matrix(validation_results, class_names, model_name='random_forest'):
    """
    Plots a heatmap of the confusion matrix for a specific model.
    """
    if model_name not in validation_results:
        print(f"Cannot plot confusion matrix: {model_name} not in validation results.")
        return
        
    cm = validation_results[model_name]['confusion_matrix']
    if cm.size == 0:
        print(f"Cannot plot confusion matrix: Matrix for {model_name} is empty.")
        return
    
    # Ensure class_names matches dimensions, or truncate
    if len(class_names) != cm.shape[0]:
        print(f"Warning: Class names ({len(class_names)}) and CM dimensions ({cm.shape[0]}) mismatch.")
        # Truncate or use generic labels if mismatch is severe
        if len(class_names) > cm.shape[0]:
            class_names = class_names[:cm.shape[0]]
        else:
            # This case is tricky, we'll just use numbers
            class_names = [f'Class {i}' for i in range(cm.shape[0])]

    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_names, yticklabels=class_names)
    plt.title(f'Confusion Matrix - {model_name}')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.show()

# Incremental learning from user interactions
class IncrementalMedicalLearning:
    def __init__(self, base_model, all_classes=None):
        if not hasattr(base_model, 'partial_fit'):
            raise TypeError(f"Model {type(base_model).__name__} does not support incremental learning (no 'partial_fit').")
        self.model = base_model
        self.interaction_buffer = []
        self.update_threshold = 100  # Update after 100 new interactions
        self.classes_ = all_classes # Must provide all possible classes for first fit
        self.initial_fit_done = False
    
    def log_user_interaction(self, symptoms, actual_diagnosis, confidence):
        # Don't log if symptoms are None (used to trigger update)
        if symptoms is not None:
            interaction = {
                'symptoms': symptoms,
                'diagnosis': actual_diagnosis,
                'confidence': confidence,
                'timestamp': datetime.now()
            }
            self.interaction_buffer.append(interaction)
        
        # Trigger model update if threshold reached
        if len(self.interaction_buffer) >= self.update_threshold:
            self.update_model()
    
    def update_model(self):
        print("\n--- Attempting Incremental Model Update ---")
        if not self.interaction_buffer:
            print("No new interactions in buffer. Skipping update.")
            return

        # Prepare training data from interactions
        # Note: This assumes 'symptoms' are already preprocessed feature vectors
        X_new = [interaction['symptoms'] for interaction in self.interaction_buffer]
        y_new = [interaction['diagnosis'] for interaction in self.interaction_buffer]
        
        # Convert to numpy arrays
        X_new = np.array(X_new)
        y_new = np.array(y_new)
        
        try:
            # Incremental training
            if not self.initial_fit_done:
                if self.classes_ is None:
                    print("Error: 'all_classes' must be provided to IncrementalMedicalLearning for the first 'partial_fit'.")
                    return
                self.model.partial_fit(X_new, y_new, classes=self.classes_)
                self.initial_fit_done = True
            else:
                self.model.partial_fit(X_new, y_new)
                
            # Clear buffer
            print(f"Model updated with {len(self.interaction_buffer)} new user interactions.")
            self.interaction_buffer = []
            
        except Exception as e:
            print(f"Error during model update: {e}")


# ============================================================================
# --- Main execution ---
# ============================================================================
if __name__ == "__main__":
    
    print("\n" + "#"*70)
    print("RUNNING: Pipeline 1 (Scikit-learn Ensemble on Mock Data)")
    print("#"*70)
    
    # Creating mock data to make the script runnable
    data = {
        'Disease': ['Common Cold', 'Flu', 'Allergy', 'Common Cold', 'Flu', 'Allergy', 'Common Cold', 'Flu', 'Gastroenteritis'],
        'Symptom1': ['runny nose', 'fever', 'sneezing', 'cough', 'body ache', 'runny nose', 'sneezing', 'fever', 'vomiting'],
        'Symptom2': ['cough', 'body ache', 'runny nose', 'runny nose', 'fatigue', 'itchy eyes', 'cough', 'cough', 'diarrhea'],
        'Symptom3': ['sneezing', 'fatigue', 'itchy eyes', 'sore throat', 'cough', np.nan, 'headache', 'headache', 'fever'],
        'Age': [25, 45, 30, 25, 50, 32, 28, 42, 35],
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male']
    }
    original_df = pd.DataFrame(data)
    
    # Create a version for graphing that has NaN placeholders
    graphing_df = original_df.copy()
    symptom_cols = [col for col in graphing_df.columns if col.startswith('Symptom')]
    graphing_df[symptom_cols] = graphing_df[symptom_cols].fillna('nan_placeholder')

    print("--- Original Mock Data ---")
    print(original_df.head())
    
    # --- Graphing from Original Data ---
    
    # 1. Example 1: Create a Visual Graph (Bar Plot)
    print("\nShowing Example 1: Visual Plot (Disease Distribution)...")
    plot_disease_distribution(original_df)

    # 2. Example 2: Create a Data Structure Graph (NetworkX)
    print("\nShowing Example 2: Network Graph (Symptom-Disease Relations)...")
    create_symptom_disease_graph(graphing_df)
    
    
    # --- ML Model Preprocessing and Training ---
    
    # 1. Preprocess the data for ML
    # --- FIX: Pass the target_column name to the preprocessor ---
    processed_df, label_encoders = preprocess_medical_data(original_df.copy(), target_column='Disease')
    
    print("\n--- Preprocessed Mock Data (Scaled and Encoded) ---")
    print(processed_df.head())

    if 'Disease' not in processed_df.columns:
        print("Stopping: 'Disease' column not found in preprocessed data. Cannot proceed with ML.")
    else:
        # 2. Separate features (X) and target (y)
        y = processed_df['Disease']
        X = processed_df.drop('Disease', axis=1)

        # 3. Split data into training and testing sets
        if len(X) < 10:
             print("\nWarning: Mock dataset is too small. Using all data for train/test.")
             X_train, X_test, y_train, y_test = X, X, y, y
        else:
             X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
        
        # --- START OF CORRECTED INDENTATION ---
        print(f"Data split: {len(X_train)} train samples, {len(X_test)} test samples.")

        # 4. Build, Train, and Evaluate Models
        models = build_diagnosis_ensemble()
        accuracies = train_and_evaluate_models(models, X_train, y_train, X_test, y_test)
        
        # 5. Run Comprehensive Validation
        print("\n--- Running Comprehensive Validation ---")
        validation_results = comprehensive_model_validation(models, X_test, y_test)
        
        # 6. Plot the new graph: Performance Dashboard
        print("\nShowing Example 3: Visual Plot (Performance Dashboard)...")
        # (This replaces the old plot_model_accuracies)
        fig = create_performance_dashboard(validation_results)
        plt.show()

        # 7. Plot Confusion Matrix for best model (e.g., Random Forest)
        print("\nShowing Example 4: Visual Plot (Confusion Matrix)...")
        class_names = label_encoders['Disease'].classes_
        plot_confusion_matrix(validation_results, class_names, model_name='random_forest')

        # 8. Calculate Confidence Scores (original function)
        print("\n--- Calculating Confidence Scores (from original function) ---")
        predictions, confidences = calculate_confidence_scores(models, X_test)
        
        if 'random_forest' in confidences:
            print("Example confidences (Random Forest) for first 5 test samples:")
            if len(confidences['random_forest']) > 5:
                print(confidences['random_forest'][:5])
            else:
                print(confidences['random_forest'])
        
        # 9. Test Incremental Learner
        print("\n--- Testing Incremental Learning ---")
        nb_model = models.get('naive_bayes')
        if nb_model:
            # We need all possible class labels (encoded) for the first partial_fit
            all_classes = np.unique(y)
            incremental_learner = IncrementalMedicalLearning(base_model=nb_model, all_classes=all_classes)
            print(f"Incremental learner initialized with {type(incremental_learner.model).__name__}.")
            
            # Simulate a new interaction (using the first test sample)
            if len(X_test) > 0:
                print("Simulating new user interaction...")
                sample_symptoms = X_test.iloc[0].values if isinstance(X_test, pd.DataFrame) else X_test[0]
                sample_diagnosis = y_test.iloc[0] if isinstance(y_test, pd.Series) else y_test[0]
                
                # The Incremental learner expects symptoms as a vector and diagnosis as an encoded label
                incremental_learner.interaction_buffer.append({
                    'symptoms': sample_symptoms,
                    'diagnosis': sample_diagnosis,
                    'confidence': 0.99, # Simulated
                    'timestamp': datetime.now()
                })
                # Set threshold low to trigger update
                incremental_learner.update_threshold = 1
                incremental_learner.log_user_interaction(None, None, None) # This log call is just to trigger the update
                
        else:
            print("Could not initialize incremental learner (Naive Bayes model not found).")
        # --- END OF CORRECTED INDENTATION ---

    print("\n" + "#"*70)
    print("ATTEMPTING: Pipeline 2 (Keras NN on 'DATASETS/Training.csv')")
    print("#"*70)
    
    # --- This will run your new Keras pipeline ---
    # It will only work if 'DATASETS/Training.csv' exists
    run_keras_training_pipeline()

    # --- New section to demonstrate streaming graph ---
    print("\n" + "#"*70)
    print("RUNNING: Pipeline 3 (Simulated Streaming Results Graph)")
    print("#"*70)
    
    print("\nNote: PySpark Streaming (Sections 4-5) runs as a continuous job.")
    print("It must be started separately (e.g., by calling ssc.start()).")
    print("We will now plot a *simulation* of its output.")

    # 1. Create some sample data, as if it came from the stream
    sample_streaming_predictions = [
        'Influenza', 'Common Cold', 'Gastroenteritis', 'Influenza',
        'Common Cold', 'Migraine', 'Influenza', 'Allergic Reaction',
        'Strep Throat', 'Common Cold', 'Gastroenteritis', 'Influenza'
    ]
    
    # 2. Plot the results
    plot_streaming_results(sample_streaming_predictions)

    # --- New section to test encryption ---
    print("\n" + "#"*70)
    print("RUNNING: Pipeline 4 (Encryption Test)")
    print("#"*70)
    
    if CRYPTOGRAPHY_AVAILABLE:
        print("Testing encryption functions...")
        try:
            key = generate_secret_key()
            iv, ciphertext, tag = encrypt_data(key, "Sensitive diagnosis: patient has diabetes type II.")

            print("  - Original:", "Sensitive diagnosis: patient has diabetes type II.")
            print("  - Encrypted:", ciphertext)
            
            decrypted = decrypt_data(key, iv, ciphertext, tag)
            print("  - Decrypted:", decrypted)
            
            if decrypted == "Sensitive diagnosis: patient has diabetes type II.":
                print("  - Status: ✓ Encryption/Decryption test PASSED.")
            else:
                print("  - Status: ✗ Encryption/Decryption test FAILED.")
        except Exception as e:
            print(f"  - Encryption test FAILED with error: {e}")
    else:
        print("Skipping encryption test, 'cryptography' library not available.")


    print("\n--- All pipelines complete. ---")
    print("--- Note: Vosk, Kivy, PySpark, and TFLite functions are defined but not executed by default. ---")

```
